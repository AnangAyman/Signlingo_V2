```python
import numpy as np, tensorflow as tf
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, log_loss)

model = tf.keras.models.load_model("models/signlingo_gru_best.h5")

# ---- Table I: architecture + parameter footprint ----
model.summary()                       # read per-layer Output Shape + Param #
print("TOTAL PARAMS:", model.count_params())

# ---- Table II: test-set metrics ----
d = np.load("test_data.npz")
X_test, y_test = d["X_test"], d["y_test"]
probs  = model.predict(X_test, verbose=0)
y_pred = probs.argmax(1)
y_true = y_test.argmax(1)
labels = list(range(probs.shape[1]))
print("Test Accuracy :", round(accuracy_score(y_true, y_pred)*100, 2), "%")
print("Test Loss     :", round(log_loss(y_true, probs, labels=labels), 4))
print("Precision (w) :", round(precision_score(y_true, y_pred, average='weighted', zero_division=0), 4))
print("Recall (w)    :", round(recall_score(y_true, y_pred, average='weighted', zero_division=0), 4))
print("F1 (w)        :", round(f1_score(y_true, y_pred, average='weighted', zero_division=0), 4))
```
